# Bar Chart
A bar chart displays the counts of distinct values in a column. Seaborn's `countplot` counts and draws in one step.

Set `column_name` to the (categorical) column you want. The `order=` argument sorts the bars - sort by the value with `sorted(...)`, or by count with `value_counts().index`. Remember to add a title and x-axis label.

```python
# set this variable to the column name you want to graph (e.g. 'genre')
column_name = 'genre'

fig, ax = plt.subplots(figsize=(14, 8))
sns.countplot(data=df, x=column_name,
              order=df[column_name].value_counts().index, ax=ax)

ax.set(xlabel="", ylabel="Count", title="")
```

# Donut Chart

A donut chart shows the same counts as a pie - each slice is one category's share of the whole - with the middle cut out. (Seaborn doesn't make pie/donut charts, so this one uses Matplotlib; the `wedgeprops` width is what makes the hole.)

Make the donut of the **same column you just charted as a bar**, then flip back and forth between the two. Which one makes it easier to put the categories in order and spot ones that are close? This is the comparison from **Lesson 5**: our eyes read the *lengths* in a bar chart far more accurately than the *angles* in a pie or donut, so a bar is almost always the clearer choice.

```python
# set this variable to the column name you want to graph (e.g. 'genre')
column_name = 'genre'

counts = df[column_name].value_counts()

fig, ax = plt.subplots(figsize=(8, 8))
ax.pie(counts, labels=counts.index, autopct='%1.1f%%',
       wedgeprops=dict(width=0.4))   # width < 1 makes the hole -> donut
ax.set(title=" ")
```

# Histogram

Histograms are used with one column of **numeric** data when there are too many distinct values to graph with a bar chart. A histogram creates 'bins' and displays the count of values that fall within each bin's range.

Choose a numeric column with many distinct values (e.g. `danceability`, `energy`, `tempo`). Adjust the bin size until you create an interesting graph. Remember to add a title and axis labels.

```python
column_name = 'danceability'

fig, ax = plt.subplots(figsize=(14, 8))
sns.histplot(data=df, x=column_name, bins=10, ax=ax)
ax.set(title=" ", xlabel=" ", ylabel="Count")
```
