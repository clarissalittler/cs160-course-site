# Scatter Plot

```python
# Create the scatter plot, specify two column names for x and y axis.
# Both columns are quantitative (numbers).
fig, ax = plt.subplots(figsize=(10, 8))
sns.scatterplot(data=df, x="Max Weight", y="Max Life Span", ax=ax)

# Add labels to the chart
ax.set(title="Max Weight vs Max Life Span",
       xlabel="Max Weight (lbs)", ylabel="Max Life Span (years)")
```

# Scatter Plot Range Chart

```python
# A range chart puts a categorical column on the y axis.
fig, ax = plt.subplots(figsize=(14, 8))
sns.scatterplot(data=df, x="Max Weight", y="Breed Group", ax=ax)

# Draw vertical lines that divide dogs into small (< 25 lbs),
# medium (25 - 55 lbs), and large (> 55 lbs) categories.
# ax.axvline() draws a vertical line, ax.axhline() draws a horizontal one.
ax.axvline(x=25, color="r", label="Small - Medium")
ax.axvline(x=55, color="g", label="Medium - Large")

ax.set(title="Max Weight per Breed Group",
       xlabel="Max Weight (lbs)", ylabel="Breed Group")
ax.legend()   # show the labels for the vertical lines
```
