```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

The data loads straight from the web - just run the cell below, there's nothing to download.

```python
# Create a dataframe from our words.csv file
df = pd.read_csv("https://drive.google.com/uc?export=download&id=11yu27f1oRyZm1awfxgcu7cLY-qKT_TvT")
```

```python
# Display the column names
df.columns
```

A crosstab chart counts how many times combinations of values appear. The example below counts word **Length** for each **Part of Speech**. Change the two column names to explore other combinations (pick columns that don't have too many distinct values, or you'll get a giant table).

# Crosstab (Cross-Tabulation) Chart

```python
# index = the rows, columns = the columns, margins=True adds totals
pd.crosstab(index=df['Part of Speech'], columns=df['Length'], margins=True)
```

# Crosstab Chart with Heatmap

A heatmap shades each cell by its value so the high counts stand out. Change `cmap` to try other colors (e.g. `"Oranges"`, `"Blues"`).

```python
counts = pd.crosstab(index=df['Part of Speech'], columns=df['Length'])

fig, ax = plt.subplots(figsize=(14, 6))
sns.heatmap(counts, annot=True, fmt='d', cmap='Greens', ax=ax)
ax.set(title="Word Length Counts per Part of Speech")
```
