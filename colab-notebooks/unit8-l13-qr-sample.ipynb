```python
# Question we are trying to answer:
# "Which breed groups are the most adopted?"
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

The data loads straight from the web - just run the cell below, there's nothing to download.

```python
# Create a dataframe from our QR-Sample.csv file
df = pd.read_csv("https://drive.google.com/uc?export=download&id=1X9anHDY1FY_sJV9_AcRpU6YuCOZN5qqi")
```

# Group by Breed Group

`groupby` collects the rows that share a Breed group, and `sum()` adds up a numeric column within each group.

```python
df.groupby('Breed group')['Total Dogs Available'].sum()
```

# Bar Chart

```python
# add up the dogs adopted within each breed group
adopted = df.groupby('Breed group', as_index=False)['Number of dogs adopted'].sum()

fig, ax = plt.subplots(figsize=(12, 7))
sns.barplot(data=adopted, x='Breed group', y='Number of dogs adopted', ax=ax)
ax.set(xlabel="Breed Group", ylabel="Dogs Adopted", title="Dogs Adopted by Breed Group")
plt.xticks(rotation=45, ha='right')   # angle the long labels so they fit
```

# Donut or Pie Chart

(Seaborn doesn't make pie/donut charts, so this one uses Matplotlib - the `wedgeprops` width is what turns the pie into a donut.)

```python
adopted = df.groupby('Breed group')['Number of dogs adopted'].sum()

fig, ax = plt.subplots(figsize=(8, 8))
ax.pie(adopted, labels=adopted.index, autopct='%1.1f%%', wedgeprops=dict(width=0.4))
ax.set(title="Share of Adoptions by Breed Group")
```

# Bar Chart with Percentages

Counts can mislead: a group might adopt out a lot of dogs simply because it *has* a lot. Dividing adoptions by dogs available gives the adoption **rate**, which is a fairer comparison.

```python
available = df.groupby('Breed group')['Total Dogs Available'].sum()
adopted   = df.groupby('Breed group')['Number of dogs adopted'].sum()

percent = (100 * adopted / available).reset_index(name='Percent Adopted')

fig, ax = plt.subplots(figsize=(12, 7))
sns.barplot(data=percent, x='Breed group', y='Percent Adopted', ax=ax)
ax.set(xlabel="Breed Group", ylabel="% Adopted", title="Adoption Rate by Breed Group")
plt.xticks(rotation=45, ha='right')
```

# Scatter Plot

Does having more dogs available go with more dogs adopted? A scatter plot puts one column on each axis so we can look for a relationship.

```python
fig, ax = plt.subplots(figsize=(10, 8))
sns.scatterplot(data=df, x="Total Dogs Available", y="Number of dogs adopted", ax=ax)

# You can add a reference line if it helps, e.g.:
# ax.axvline(x=20, color="r", label="my cutoff"); ax.legend()

ax.set(title="Total Dogs Available vs Number of Dogs Adopted",
       xlabel="Total Dogs Available", ylabel="Number of Dogs Adopted")
```
