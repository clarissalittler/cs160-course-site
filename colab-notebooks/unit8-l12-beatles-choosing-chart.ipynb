# Scatter Plot

Scatter plots show the relationship between two columns. Choose 2 numeric columns (or one text and one numeric) and fill in the column names below. Remember to add a title.

```python
# fill in two column names for x and y (e.g. "energy" and "danceability")
fig, ax = plt.subplots(figsize=(14, 8))
sns.scatterplot(data=df, x="", y="", ax=ax)

# Uncomment to draw reference lines, e.g.:
# ax.axhline(y=0.5, color="r")

ax.set(title="", xlabel="", ylabel="")
```

# Crosstab (Cross-Tabulation) Chart

A crosstab chart displays counts of the distinct values in one column mapped against another. Only use columns that have a small number of distinct values, or you'll generate a huge table!

The example below crosses `genre` with `key`. Change the two column names to your own choice. Change `cmap` to try other heatmap colors.

```python
counts = pd.crosstab(index=df['genre'], columns=df['key'])

fig, ax = plt.subplots(figsize=(14, 6))
sns.heatmap(counts, annot=True, fmt='d', cmap='Oranges', ax=ax)
ax.set(title="Count of key per genre")
```

```python
# The same crosstab as a plain table with totals:
pd.crosstab(index=df['genre'], columns=df['key'], margins=True)
```
