# Cross-tabulation (Crosstab) Chart

`pd.crosstab()` counts the combinations. `index=` is the rows, `columns=` is the columns, and `margins=True` adds the row/column totals.

```python
# the count of each Max Life Span value per Breed Group, with totals
pd.crosstab(index=df['Breed Group'], columns=df['Max Life Span'], margins=True)
```

# Crosstab Chart with Heatmap

A **heatmap** shades each cell by its value, so the high counts jump out. Seaborn's `heatmap()` draws one directly from the crosstab. `annot=True` writes the number in each cell; change `cmap` to try different colors (e.g. `"Greens"`, `"Blues"`).

```python
counts = pd.crosstab(index=df['Breed Group'], columns=df['Max Life Span'])

fig, ax = plt.subplots(figsize=(14, 6))
sns.heatmap(counts, annot=True, fmt='d', cmap='Oranges', ax=ax)
ax.set(title="Maximum Life Span Counts per Breed Group")
```
