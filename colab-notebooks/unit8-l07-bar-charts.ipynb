# Bar Chart

```

# Create a new dataframe called bar_df with each ‘Max Life Span’ value and the number of times (counts) that value appears in the column
bar_df = df['Max Life Span'].value_counts().to_frame('count').reset_index()

# Sort the dataframe by the ‘Max Life Span’ column
bar_df = bar_df.sort_values('Max Life Span')

# Plot the dataframe and use the column ‘index’ for x-axis
bar_df.plot.bar(x='Max Life Span', figsize=(14, 8))

#Add labels to the chart
plt.xlabel("# Years")
plt.ylabel("Count")
plt.title("Count of Max Life Spans")
```
