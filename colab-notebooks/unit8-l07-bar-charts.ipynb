# Bar Chart

A bar chart counts how many times each value appears in a column. Seaborn's `countplot` does the counting *and* the drawing in one step - we just hand it the column. (We use `order=` to put the bars in order along the x-axis.)

```python
fig, ax = plt.subplots(figsize=(14, 8))

# count how many breeds have each 'Max Life Span' value
sns.countplot(data=df, x="Max Life Span",
              order=sorted(df["Max Life Span"].dropna().unique()), ax=ax)

# Add labels to the chart
ax.set(xlabel="# Years", ylabel="Count", title="Count of Max Life Spans")
```
