# Do something again and again and again and...
The next concept to toy with is *iteration*, which is the technical term for saying "I want to do the same thing a bunch of times."

One of the basic lessons of being a programmer is that if you see something repeated a bunch of times then there's a way to make it simpler. We made our square, earlier, but we had to type out forward and turn in the exact same way over and over.

Here's the basic Python idiom for how to do something multiple times:

```python
initializeTurtle()

for i in range(4): # do this for every number from 0 - 3 (0, 1, 2, 3)
  # print(i)
  forward(50)
  right(90)
```

There's a couple of pieces here to talk about. First, you'll notice that the line for the iteration starts with `for i in range(4):` So the `for` starts off the iteration and it isn't optional, the next part the `i` is a variable that will count how many times we've Done The Thing, `in range(4)` means that we're saying to Do The Thing four times total.

The part that's visually the weird is that this line ends with a ":" and then the next two lines are indented.

Basically, the next two lines are indented in because that's the way we tell Python "these lines of code are what need to be repeated". Once the Python interpreter sees its first unindented lines it knows that these aren't in the loop anymore.

Here's an example to make that more concrete

```python
# what lines are actually repeated?

initializeTurtle()

for i in range(10):
  forward(5)
  right(10)

forward(100)
```

```python
# We can even use the count variables inside the loop

initializeTurtle()

for i in range(30):
  forward(5*i)
  right(3*i)
```

If you run that code you'll be able to see that the longer it goes the longer each step is and the sharper each turn is.

That's because for the entire execution of this loop `i` takes values from 0 to 29. Yes, that's right, *0* to *29*. Counting in programming almost always (with the exception of the language Lua and a couple of others) starts with 0 not 1.

You might also be wondering if the counting variable always has to be `i` and the answer is no! But it's kind of programming convention to use `i` and `j` as your indexes for iteration, which has the advantage of letting you skim code a little easier.

So unless you have a good reason to use a different variable name, I'd definitely just stick with `i` and `j`.

Here's another example that involves combining iteration with variables to create a tweakable polygon drawing program. Note that you can change the number of sides as well as the leg length in order to change the polygon.

Question: How should you change the `numSides` variable in order to (approximately) draw a circle? You might need to change the leg length to stay on the screen!

```python
# abstracting angle
initializeTurtle()

numSides = 4
leg = 40

# make it go faster
speed(7)

for i in range(numSides):
  forward(leg)
  right(360/numSides)
```

# New verbs: penup, pendown, goto

If we want to make more complicated pictures we're going to need to be able to move around the canvas without drawing. We need to be able to, metaphorically, lift our pen up off the canvas and then put it down later.

Well luckily for us this graphics library includes three more *verbs* for us to use: `penup`, `pendown`, and `goto`.

`goto` is going to be our first "verb" that takes more than one argument. Check out the following code:

```python
initializeTurtle()

speed(1)

goto(0,0)

goto(200,200)

goto(0,200)

goto(200,0)
```

(0,0) is the upper-left corner of the screen
y gets bigger as you move down
x gets bigger as you move right

`window_width()` returns the width

`window_height()` returns the height

Okay so we'll try an example of using these things to write code that works no matter how big your canvas is

```python
initializeTurtle()

# let's say the canvas is actually 800x500
w = window_width()
h = window_height()

color('MediumBlue')

goto(2,2)

goto(2,h-2)

goto(w-2,h-2)

goto(w-2,2)

goto(w/2,h/2)
```

So the problem with the above is that as you move between spaces you can see that *you're still drawing*. Really what you need is the ability to "pick up the pen" as you move and "put the pen down" when you're ready to draw again.

Well! It turns out that there actually are verbs for this: `penup()` and `pendown()`

```python
initializeTurtle()

penup()

goto(0,0)

pendown()

goto(100,100)
```

Finally, here's an example of drawing two hexagons separately from each other using the for-loop tricks we've used before *and* judicious use of penup and pendown

Huh. Kinda stinks to have to repeat that for-loop to draw the same thing, doesn't it? We'll see how to fix that in the next notebook.

```python
# example

initializeTurtle()

numSides = 6
leg = 40

penup()
goto(2*leg,2*leg)
pendown()

for i in range(numSides):
  forward(leg)
  right(360/numSides)

penup()
goto(4*leg,4*leg)
pendown()

for i in range(numSides):
  forward(leg)
  right(360/numSides)
```
