# File functions

Adapt the field names and conversions to your schema.

````python
import csv

FILENAME = "records.csv"
FIELD_NAMES = ["name", "category", "amount", "notes"]

def load_records(filename):
    records = []
    problems = []

    try:
        with open(filename, "r", encoding="utf-8", newline="") as data_file:
            reader = csv.DictReader(data_file)
            for row_number, row in enumerate(reader, start=2):
                try:
                    amount = float(row["amount"])
                except ValueError:
                    problems.append(f"Row {row_number}: amount is not numeric")
                    continue

                records.append({
                    "name": row["name"].strip(),
                    "category": row["category"].strip(),
                    "amount": amount,
                    "notes": row.get("notes", "").strip(),
                })
    except FileNotFoundError:
        print(f"No saved {filename}; starting with an empty collection.")

    return records, problems

def save_records(filename, records):
    with open(filename, "w", encoding="utf-8", newline="") as data_file:
        writer = csv.DictWriter(data_file, fieldnames=FIELD_NAMES)
        writer.writeheader()
        writer.writerows(records)
````

# Main menu

Save whenever records change.

````python
records, problems = load_records(FILENAME)
print(f"Loaded {len(records)} records; skipped {len(problems)} malformed rows.")
for problem in problems:
    print(problem)

choice = ""
while choice != "5":
    show_menu(len(records))
    choice = input("Choose 1 through 5: ").strip()

    if choice == "1":
        changed = add_record(records)
        if changed:
            save_records(FILENAME, records)
            print(f"Saved {len(records)} records to {FILENAME}.")
    elif choice == "2":
        display_records(records)
    elif choice == "3":
        filter_records(records)
    elif choice == "4":
        show_summary(records)
    elif choice == "5":
        print("Goodbye.")
    else:
        print("That is not a menu option. Your records have not changed.")
````
