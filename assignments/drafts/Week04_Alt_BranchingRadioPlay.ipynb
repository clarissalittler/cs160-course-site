# Defining a reusable scene function

```python
def scene():
    print()
    print("SCENE BEGINS")
    print("A room waits in complete silence.")
    print("SCENE ENDS")

scene()
scene()
```

A function definition begins with `def`, followed by the function name, parentheses, and a colon. The indented lines are the function's **body**.

When Python reaches `def scene():`, it learns what the new verb means. It does not run the body yet. Each later `scene()` is a function **call**, and each call runs the body once.

This is similar to a `for` loop because indentation marks a block of related instructions. The difference is that a loop runs its block as part of the loop, while a function body runs whenever the function is called.

# Parameters: changing the details

The first `scene()` function always printed the same room. A **parameter** is a local name that receives a value from a function call. Parameters let the repeated structure stay consistent while the content changes.

```python
def scene(place):
    print()
    print("SCENE")
    print("Location:", place)

scene("an empty train platform")
scene("a crowded all-night diner")
scene("the school library after closing")
```

Inside the first call, `place` refers to `"an empty train platform"`. The second call supplies a different value.

Let us add more parameters one at a time:

```python
def scene(place, sound, event):
    print()
    print("SCENE")
    print("Location:", place)
    print("Sound:", sound)
    print("Event:", event)

scene(
    "an empty train platform",
    "a fluorescent light buzzing",
    "a suitcase moves by itself"
)

scene(
    "the roof",
    "wind through antennas",
    "a paper map points north by itself"
)
```

The function has three parameters: `place`, `sound`, and `event`. Each function call supplies three concrete values in the same order. Those concrete values are sometimes called **arguments**.

The second call does not duplicate the labels and layout. If we decide that every scene needs an `Event:` label or a blank line, we can change the function once instead of finding every scene in the program.

# Generalizing the function further

What other repeated work belongs inside the function? A title and speaker could also vary.

```python
def scene(title, place, sound, speaker, dialogue):
    print()
    print("SCENE TITLE:", title)
    print("Location:", place)
    print("Sound:", sound)
    print(f"{speaker} says: {dialogue}")

scene(
    "THE PLATFORM",
    "an empty station",
    "a train approaching very slowly",
    "NARRATOR",
    "The last train was not on the schedule."
)

scene(
    "THE CARRIAGE",
    "the final train car",
    "wheels clicking on the track",
    "CONDUCTOR",
    "Tickets, memories, or unfinished business, please."
)
```

This function is more general, but more parameters are not automatically better. Include a parameter when different calls genuinely need different values. A clear function with three useful parameters is better than a confusing function with ten.

# Asking the reader to choose

The `input()` function pauses the program and returns the text the reader enters.

```python
choice = input("Do you enter the elevator or take the stairs? ")
print("You entered:", choice)
```

To make the story branch, use an `if` statement:

```python
choice = input("Type elevator or stairs: ")

if choice == "elevator":
    print("The elevator button is already lit.")
else:
    print("The stairs go down, even though you started on the first floor.")
```

The expression `choice == "elevator"` is a comparison. It asks a yes-or-no question.

Be careful with the equals signs:

- `choice = input(...)` assigns a value to `choice`.
- `choice == "elevator"` compares two values.

If the comparison is true, Python performs the indented `if` block and skips the `else` block. If it is false, Python skips the first block and performs the `else` block.

# Making input more forgiving

People may type `Radio`, `radio `, or ` RADIO`. The string methods `.strip()` and `.lower()` help us compare all of those consistently.

```python
choice = input("Choose radio or letter: ")
choice = choice.strip().lower()

if choice == "radio":
    print("The radio says your name, then apologizes.")
else:
    print("You leave the radio untouched.")
```

`.strip()` removes extra spaces from the beginning and end. `.lower()` converts letters to lowercase. The updated string is assigned back to `choice`.

# More than two paths with `elif`

The previous example treats every answer other than `radio` in the same way. `elif`, short for “else if,” gives us another explicit comparison.

```python
choice = input("Choose radio or letter: ")
choice = choice.strip().lower()

if choice == "radio":
    print("The radio says your name, then apologizes.")
elif choice == "letter":
    print("The letter is addressed to tomorrow morning.")
else:
    print("You choose neither, which is also a choice.")
```

Python checks the branches from top to bottom and performs only the first one whose comparison is true. The final `else` is a fallback for any value not handled above.

## A tiny branching scene

```python
def announce(chapter, text):
    print()
    print("CHAPTER:", chapter)
    print(text)

announce("OPENING", "You are alone in the school after dark.")

choice = input("Do you follow the music or the footsteps? ")
choice = choice.strip().lower()

if choice == "music":
    announce("THE MUSIC ROOM", "Every instrument is playing one note, waiting for you.")
elif choice == "footsteps":
    announce("THE HALLWAY", "The footsteps stop whenever you stop.")
else:
    announce("THE PAUSE", "You wait. Both sounds begin moving toward you.")

announce("ENDING", "Somewhere, a bell rings once.")
```

Notice that `announce()` is called before the condition, inside all three branches, and after the condition. Only one middle call runs on a particular playthrough, but the function itself handles the same repeated chapter-formatting job each time.

# Combining conditions with `and`

Sometimes a scene depends on two facts at once. The `and` operator is true only when both comparisons are true.

```python
has_key = True
trust = 3

if has_key == True and trust >= 2:
    print("The locked door opens, and your companion follows.")
elif has_key == True and trust < 2:
    print("The door opens, but your companion stays behind.")
else:
    print("The lock does not move.")
```

`True` and `False` are Boolean values: Python's built-in values for yes/no facts. The comparison `has_key == True` could also be written more simply as `has_key`, but the longer form makes the first example easier to read aloud.

# Combining alternatives with `or`

Use `or` when either one of two complete comparisons should take the same path.

```python
choice = "open the envelope"

if choice == "open" or choice == "open the envelope":
    print("The paper contains a map of streets that do not exist yet.")
else:
    print("The envelope remains sealed.")
```

A common mistake is writing `if choice == "open" or "open the envelope":`. That does not compare `choice` a second time. Write both complete comparisons as shown.

# State that changes as the story runs

A variable can remember what earlier scenes did. This remembered information is often called the program's **state**.

```python
tension = 0

print("Opening tension:", tension)

choice = "basement"
if choice == "basement":
    tension = tension + 2
else:
    tension = tension + 1

print("Tension after the first choice:", tension)
```

`tension = tension + 2` looks strange at first. Python evaluates the right side using the old value, then stores the new result back in the same variable.

We can keep a value inside boundaries, much as a moving object can reverse before leaving a canvas:

```python
tension = 6
change = 2

for i in range(6):
    tension = tension + change

    if tension >= 10 or tension <= 0:
        change = -change

    print(f"Scene {i + 1} tension: {tension}")
```

The `or` condition catches either boundary. Replacing `change` with `-change` reverses whether tension rises or falls. You do not need this pattern for the assignment, but it demonstrates that conditions can respond to changing state, not only to typed answers.

# A complete branching radio-play example

Read the example before running it. Find the definition, each call, the normalized input, the three-way branch, and the final scene that all paths share.

```python
def scene(title, place, sound, event):
    print()
    print("SCENE TITLE:", title)
    print("Location:", place)
    print("Sound:", sound)
    print("What happens:", event)

name = input("What is the main character's name? ")

scene(
    "OPENING",
    "a city bus after midnight",
    "soft rain and old brakes",
    f"{name} notices that every passenger is holding the same envelope"
)

choice = input("Does the character open the envelope or ask the driver? ")
choice = choice.strip().lower()

if choice == "open" or choice == "open the envelope":
    scene(
        "THE ENVELOPE",
        "seat 12",
        "paper tearing",
        "inside is a map of streets that do not exist yet"
    )
elif choice == "driver" or choice == "ask the driver":
    scene(
        "THE DRIVER",
        "the front of the bus",
        "a turn signal ticking",
        "the driver says this route only appears when someone is lost"
    )
else:
    scene(
        "THE PAUSE",
        "the aisle",
        "everyone waiting",
        "doing nothing makes the envelopes begin to hum"
    )

scene(
    "ENDING",
    "the next stop",
    "doors folding open",
    f"{name} steps out and recognizes a place they have never been"
)
```

The output is intentionally explicit: sound is labeled instead of being implied by punctuation, and location changes are stated instead of being represented by visual position.

# When Python objects: a debugging guide

- `SyntaxError` near `def`, `if`, `elif`, or `else` often means a colon is missing.
- `IndentationError` means Python cannot determine which lines belong to a function or branch.
- `NameError` often means a function or variable was misspelled, or its definition cell was not run.
- `TypeError` about missing arguments means a function call does not supply a value for every required parameter.
- If every input takes the fallback branch, print `choice` after normalization and compare its exact text with your conditions.
- If a later scene forgets an earlier event, store that event in a variable outside the function.
