# Variables: names for reusable pattern pieces

If you are building a repeated work, you will probably reuse words. A **variable** stores a value under a name so you can refer to it later.

```python
sound = "tap"
image = "window"
mood = "electric"

print(sound)
print(image)
print(mood)
```

The single equals sign is the **assignment operator**. Read `sound = "tap"` as “store the string `"tap"` under the name `sound`.” When a later instruction uses `sound`, Python looks up its current value.

```python
sound = "hiss"

print(sound)
print("The kettle says", sound)
print("The old radio answers", sound)
```

Change only the first line and run the cell again. Both later lines should change. Naming the repeated element once makes experimentation easier and prevents inconsistent edits.

Variables can also store numbers:

```python
beats = 4
verses = 3

print("Beats per verse:", beats)
print("Number of verses:", verses)
print("Total beats:", beats * verses)
```

The multiplication in `beats * verses` is an **expression**. Python computes it before `print()` runs.

# Strings can be joined and repeated

Python's `+` operator joins strings. Its `*` operator repeats a string a specified number of times.

```python
first_word = "soft"
second_word = "machine"

print(first_word + " " + second_word)
print("tap " * 4)
print((first_word + " ") * 3 + second_word)
```

Spaces inside strings matter. Compare `"tap" * 4` with `"tap " * 4`. Neither is wrong, but they produce different text.

Use punctuation or capitalization for style if you like, but make sure the words themselves carry the meaning. A repeated row of symbols may sound confusing through a screen reader; a label such as `REFRAIN` or `VERSE 2` makes the structure explicit.

# F-strings: placing variables inside a line

An **f-string** lets you write a sentence with clearly marked blanks. Put `f` before the opening quotation mark, then put variable names inside curly braces.

```python
sound = "tap"
image = "window"
mood = "electric"

line = f"The {image} goes {sound}, and the room feels {mood}."
print(line)
```

Python evaluates each expression inside braces and inserts its value. Change one variable and run the cell again to hear how the whole line changes.

```python
object_name = "umbrella"
attempt = 2

print(f"Attempt {attempt}: the {object_name} refuses to open.")
print(f"Attempt {attempt + 1}: the {object_name} opens indoors.")
```

The braces can contain a small expression such as `attempt + 1`, not only a variable name.

# Asking the reader for words

The built-in `input()` function pauses the program, presents a prompt, and returns the text the reader enters.

```python
noun = input("Give me a noun: ")
verb = input("Give me a verb ending in -ing: ")
place = input("Give me a place: ")

print()
print(f"In {place}, the {noun} is {verb}.")
```

The answer from each prompt is a string. The first line stores its answer in `noun`, the second in `verb`, and the third in `place`.

The empty `print()` adds a blank line between the questions and the finished piece. Blank space may help some readers, but it should not be the only cue. A spoken heading is clearer:

```python
speaker = input("Who is speaking? ")
message = input("What do they repeat? ")

print()
print("MONOLOGUE BEGINS")
print(f"{speaker} says: {message}")
print("MONOLOGUE ENDS")
```

# Do something again and again and again

**Iteration** is the programming term for repeating a process. It is useful whenever a work has a refrain, beat, countdown, escalation, list, or recurring sentence.

We could repeat a line manually:

```python
refrain = "The signal is still there."

print(refrain)
print(refrain)
print(refrain)
print(refrain)
```

But duplicated code is hard to revise and obscures the underlying rule. A `for` loop states the pattern directly:

```python
refrain = "The signal is still there."

for i in range(4):
    print(refrain)
```

Read the loop heading as “for each value of `i` in a range of four values, perform the indented instructions.”

Three details are important:

1. The loop starts with `for` and ends its heading with a colon.
2. `range(4)` causes four repetitions.
3. Indentation marks the block that repeats. Python conventionally uses four spaces.

## Which lines are inside the loop?

Predict how often each sentence will print, then run the code.

```python
for i in range(3):
    print("The light blinks.")
    print("Someone writes it down.")

print("The test is over.")
```

The two indented lines repeat three times. The final line is not indented, so it runs once after the loop. The change in indentation tells Python that the repeated block has ended.

# Letting a pattern change over time

During `range(5)`, the loop variable `i` receives the values `0`, `1`, `2`, `3`, and `4`. Python commonly begins counting at zero.

```python
for i in range(5):
    print("Counter value:", i)
    print("Human-friendly line number:", i + 1)
```

The counter name does not have to be `i`, although `i` is conventional for a simple counter. We can use its changing value as part of the poem.

```python
object_name = "small machine"
action = "tries to remember"

for i in range(5):
    print(f"Attempt {i + 1}: the {object_name} {action}.")
```

Or we can use it to make every line longer:

```python
word = "echo"

for i in range(6):
    repetition_count = i + 1
    print(f"Echo level {repetition_count}: " + (word + " ") * repetition_count)
```

This is a “shape” made from language: each pass follows the same rule, but the counter changes the result.

# Combining the pieces: a complete pattern poem machine

The next example combines top-to-bottom execution, input, variables, f-strings, string repetition, and a loop. Before you run it, identify what remains the same in every verse and what changes.

```python
title = input("Title: ")
speaker = input("Who is speaking? ")
place = input("Where are they? ")
object_name = input("Name an important object: ")
action = input("Give an action ending in -ing: ")
refrain = input("Give a short repeated phrase: ")

print()
print("PATTERN POEM BEGINS")
print(title.upper())
print(f"Spoken by {speaker}")

for i in range(6):
    verse_number = i + 1
    print()
    print(f"VERSE {verse_number}")
    print(f"In {place}, {speaker} sees the {object_name} {action}.")
    print((refrain + " ") * verse_number)

print()
print("PATTERN POEM ENDS")
```

This sample is intentionally simple so its structure is easy to inspect. Your assignment should have its own idea and wording rather than merely replacing the six inputs.

# When Python objects: a short debugging guide

Errors are a normal part of programming. Read the last line of the error message first.

- `SyntaxError` often means a closing quote, parenthesis, or loop colon is missing.
- `IndentationError` means Python cannot determine which lines belong to the loop.
- `NameError` often means a variable name is misspelled or the cell that creates it was not run.
- If `{speaker}` appears literally, make sure the opening quote has an `f` immediately before it.
- If words run together, include a space inside the string you repeat: `"echo " * 3`.
- If the loop runs the wrong number of times, inspect the value passed to `range()`.
