# Variables: naming values so we can reuse them

Retyping the same text repeatedly is annoying, and it makes revision harder. A **variable** lets us give a value a name.

```python
title = "THE LATE-NIGHT GUIDE TO MISSING SOCKS"
topic = "laundry portals"
warning = "Do not trust the warm towel."

print(title)
print(topic)
print(warning)
```

Read `title = "THE LATE-NIGHT GUIDE TO MISSING SOCKS"` as “store this string under the name `title`.” In programming, a single equals sign performs **assignment**. It stores the value on the right in the variable named on the left.

Variable names do not appear in the output by themselves. Python replaces each name with its current value when that line runs.

The real benefit is that one change can affect several lines:

```python
featured_object = "left glove"

print("ARCHIVE ENTRY")
print("Object:", featured_object)
print("We interviewed the", featured_object)
print("Please return the", featured_object, "before midnight.")
```

Change `featured_object` once and run the cell again. All three references should change together. This is the same computational idea as naming a length or angle once and reusing it throughout a drawing.

## Numbers and expressions

Variables can store numbers as well as strings. Python can evaluate arithmetic expressions before printing the result.

```python
issue_number = 3
pages = 4
copies = 5

print("Issue number:", issue_number)
print("Pages in each copy:", pages)
print("Total pages to print:", pages * copies)
```

Here `pages * copies` is an **expression**. Python looks up both values, multiplies them, and gives the result to `print()`.

Strings also support useful operations. The `+` operator joins strings, while `*` repeats one.

```python
word = "hush"

print("Please " + word + ".")
print((word + " ") * 4)
print("-" * 40)
```

The last line makes a divider. A divider can help organize output, but do not use spacing or punctuation as the *only* way to communicate meaning. A heading such as `FIELD NOTE` remains understandable when read aloud.

# F-strings: putting variables into sentences

An **f-string** is a convenient way to place variables inside text. Put the letter `f` before the opening quote, and put each variable name inside curly braces.

```python
place = "the basement"
object_name = "a blue sock"
problem = "it started humming"

print(f"In {place}, I found {object_name}.")
print(f"The problem was obvious: {problem}.")
```

The braces mean “evaluate what is inside here and insert its value.” Compare the output after changing only `place`.

Triple quotation marks let a string continue across several lines. This is useful when a zine section contains a paragraph.

```python
page = f"""
FIELD NOTE
Location: {place}
Discovery: {object_name}
Complication: {problem}
"""

print(page)
```

# Asking the reader for ingredients

So far, the author of the program decides every word in advance. The built-in `input()` function lets the person running the program contribute a value.

```python
hero = input("Name a person, creature, or object: ")
place = input("Name a place: ")
strange_item = input("Name a strange object: ")

print()
print(f"{hero} entered {place} carrying {strange_item}.")
```

When Python reaches `input()`, the program pauses. It displays the prompt and waits for the reader to type an answer. That answer is returned as a string and stored in the variable.

Notice the empty `print()`. With no value between its parentheses, it prints a blank line. Here the blank line separates the questions from the finished artifact. Again, headings and labels—not blank space alone—should communicate the structure.

# Do something again and again and again

The next concept is **iteration**, the programming term for repeating a process. Repetition is useful in a zine: recurring slogans, numbered field notes, escalating warnings, repeated sound effects, and lists all create structure.

We could repeat a refrain by writing the same line four times:

```python
refrain = "CHECK THE POCKETS"

print(refrain)
print(refrain)
print(refrain)
print(refrain)
```

That works, but it is repetitive code. A `for` loop expresses the pattern directly:

```python
refrain = "CHECK THE POCKETS"

for i in range(4):
    print(refrain)
```

Read the first line as “for each value of `i` in this range of four values, do the indented instructions.” The colon ends the loop heading. The indentation tells Python which instructions belong to the loop.

Python is strict about indentation. The conventional indentation is four spaces. In Colab, pressing Tab at the start of a line will usually insert the correct indentation.

## Which lines repeat?

Before you run this cell, predict how many times each sentence will appear.

```python
for i in range(3):
    print("A small door appears.")
    print("Someone knocks from the other side.")

print("The hallway becomes quiet.")
```

Both indented lines run three times. The unindented final line runs once, after the loop finishes. An unindented line tells Python that the loop's block has ended.

# Using the loop counter

The variable `i` changes during the loop. `range(5)` supplies the values `0`, `1`, `2`, `3`, and `4`. Programming languages commonly begin counting at zero.

```python
for i in range(5):
    print("Python's counter is", i)
    print("A reader-friendly item number is", i + 1)
```

The counter does not have to be called `i`, but `i` is a common short name when it merely counts repetitions. Use a more descriptive name when the values have a meaning of their own.

We can use the counter to make the output change over time:

```python
dramatic_word = "NO"

for i in range(5):
    repetition_count = i + 1
    print(f"Warning {repetition_count}: " + (dramatic_word + " ") * repetition_count)
```

The first pass uses `i = 0`, so `repetition_count` is `1`. On every later pass it grows by one. The loop has a repeated rule, but the result is not identical each time.

# Combining the pieces: a complete small zine

This example combines execution order, variables, strings, input, expressions, and a loop. Read it from top to bottom before running it. Identify which lines collect data, which lines print the frame of the zine, and which lines repeat.

```python
title = input("Title for the mini-zine: ")
creator = input("Who made it? ")
place = input("Where does it take place? ")
object_name = input("Name an important object: ")
warning = input("Write a warning: ")
refrain = input("Write a short repeated line: ")

divider = "=" * 50

print()
print(divider)
print(title.upper())
print(f"by {creator}")
print(divider)

print("FIELD NOTE")
print(f"Welcome to {place}.")
print(f"Today's featured object is {object_name}.")
print(f"Please remember: {warning}")

print()
print("REPEATED NOTICE")
for i in range(4):
    print(f"Notice {i + 1}: {refrain}")

print()
print("END OF ZINE")
```

# When Python objects: a short debugging guide

Errors are normal evidence that you are making something new. Read the last line of an error message first; it often gives the clearest clue.

- `SyntaxError` often means a quote, parenthesis, or colon is missing.
- `IndentationError` means Python cannot understand which lines belong to a loop.
- `NameError` often means a variable was misspelled or used before it received a value.
- If `{variable_name}` prints literally, check that the string begins with `f`.
- If the loop runs the wrong number of times, check the number inside `range()`.
