# Week 4 Activity — Make a Personality Quiz

## What's this all about?

“Which Twin Peaks character are you?” “What kind of bread matches your soul?” “Which food cart represents your inner life?” Personality quizzes are a staple of the web: a little ridiculous, fun to pass around, and made from exactly the ideas in this unit.

Think about what a quiz actually does. It presents a question, in the same format, over and over. That's a repeated job, so we'll give it a name with a **function**. Then every answer forces the program to **decide** what to do with the score, and one last decision picks which result to announce. By the end you'll have a small program a classmate can take, and a result to share.

Keep the theme playful or actually useful. Don't present the result as any kind of medical, psychological, or scientific diagnosis, and don't sort classmates by protected traits.

## First, one question the long way

This works. But look at how much of it's the *format* of a question, and how little is the question itself:

```python
print()
print("QUESTION")
print("Pick a place to spend an afternoon.")
print("A: A busy public square")
print("B: A quiet library corner")
answer = input("Type A or B: ").strip().upper()
print("You chose:", answer)
```

If we copy that whole block for every question, then changing the format means changing every copy. The shared job deserves a name.

## Define a reusable question function

```python
def ask_ab(question, option_a, option_b):
    print()
    print("QUESTION")
    print(question)
    print("A:", option_a)
    print("B:", option_b)
    answer = input("Type A or B: ")
    return answer.strip().upper()
```

When Python reaches `def`, it learns the new verb but doesn't ask anything yet. The body runs when we **call** `ask_ab(...)`, and the three parameters take different values on each call. `return` hands the cleaned-up answer back to the program so we can do something with it.

Try two calls:

```python
first_answer = ask_ab(
    "Pick a place to spend an afternoon.",
    "A busy public square",
    "A quiet library corner"
)

second_answer = ask_ab(
    "Pick a project.",
    "Organize a neighborhood festival",
    "Restore an old map"
)

print("First answer:", first_answer)
print("Second answer:", second_answer)
```

We wrote the asking code once, and it did the job twice with different content. That's the whole point of a function.

## Let each answer change a score

Set up a score before asking anything. Then after each call, use a conditional to decide what the answer does to it:

```python
score = 0

answer = ask_ab(
    "Your ideal Friday night?",
    "Spontaneous plans with a crowd",
    "A project and a comfortable chair"
)

if answer == "A":
    score = score + 1
elif answer == "B":
    score = score - 1
else:
    print("I did not recognize that answer, so the score did not change.")

print("Score so far:", score)
```

The `else` branch is there so an unexpected answer gets noticed instead of silently counting as B. A more complete version would use a `while` loop to ask again, which is a nice extension once you've done Lesson 5.

## Ask several questions

Here's the full scoring pattern. The same function gets called three times, and each conditional updates the same score.

```python
score = 0

answer = ask_ab(
    "A mysterious door appears. You:",
    "Open it immediately",
    "Take notes before touching it"
)
if answer == "A":
    score += 1
elif answer == "B":
    score -= 1

answer = ask_ab(
    "Choose a tool.",
    "A megaphone",
    "A pocket notebook"
)
if answer == "A":
    score += 1
elif answer == "B":
    score -= 1

answer = ask_ab(
    "A plan changes at the last minute. You:",
    "Improvise and invite everyone along",
    "Find a quiet place to redesign it"
)
if answer == "A":
    score += 1
elif answer == "B":
    score -= 1

print("Final score:", score)
```

## Announce a result

Now a final branch picks the result. The labels are made up, and the descriptions say what the quiz author means by them, which is the honest way to do it.

```python
if score >= 2:
    print("RESULT: Signal Fire")
    print("You like momentum, people, and making an idea visible.")
elif score <= -2:
    print("RESULT: Deep Archive")
    print("You like careful observation, focus, and building an idea over time.")
else:
    print("RESULT: Bridge")
    print("You move between spontaneous and reflective ways of working.")
```

Test the boundaries: scores of 2, 1, -1, and -2 should land in the categories you meant. Those thresholds are choices you made, not facts hiding in the data, and there's nothing wrong with that as long as you know it.
