# Week 4 Activity — Make a Personality Quiz

## What's this all about?

“Which Twin Peaks character are you?” “What kind of bread matches your soul?” “Which food cart represents your inner life?” Personality quizzes are a staple of the web: a little ridiculous, fun to pass around, and made from exactly the ideas in this unit.

A quiz needs to perform one repeated job—present a question in a consistent format—so we will name that job with a **function**. Each answer forces the program to make a **decision** about the score. A final decision chooses which result to announce. By the end, you will have a small program another person can take and a result worth sharing with the class.

Keep the theme playful or genuinely useful. Do not present the result as a medical, psychological, or scientific diagnosis, and do not sort classmates using protected traits.

## First, one question the long way

This works, but it mixes the question's reusable presentation with one particular question:

```python
print()
print("QUESTION")
print("Pick a place to spend an afternoon.")
print("A: A busy public square")
print("B: A quiet library corner")
answer = input("Type A or B: ").strip().upper()
print("You chose:", answer)
```

If we copy that whole block for every question, a wording change has to be made in every copy. The shared job deserves a name.

## Define a reusable question function

```python
def ask_ab(question, option_a, option_b):
    print()
    print("QUESTION")
    print(question)
    print("A:", option_a)
    print("B:", option_b)
    answer = input("Type A or B: ")
    return answer.strip().upper()
```

When Python reaches `def`, it learns the new verb but does not ask anything yet. The body runs when we **call** `ask_ab(...)`. The three parameters receive different arguments on each call. `return` hands the normalized answer back to the program.

Try two calls:

```python
first_answer = ask_ab(
    "Pick a place to spend an afternoon.",
    "A busy public square",
    "A quiet library corner"
)

second_answer = ask_ab(
    "Pick a project.",
    "Organize a neighborhood festival",
    "Restore an old map"
)

print("First answer:", first_answer)
print("Second answer:", second_answer)
```

The asking code was written once, but the function performed that job twice with different content.

## Let each answer change a score

Create a score before asking questions. After each call, use a conditional to decide how the answer changes it:

```python
score = 0

answer = ask_ab(
    "Your ideal Friday night?",
    "Spontaneous plans with a crowd",
    "A project and a comfortable chair"
)

if answer == "A":
    score = score + 1
elif answer == "B":
    score = score - 1
else:
    print("I did not recognize that answer, so the score did not change.")

print("Score so far:", score)
```

The `else` branch makes the current limitation visible instead of silently treating every unexpected answer as B. A more complete program could use a `while` loop to ask again; that is a good optional extension after Lesson 5.

## Ask several questions

Here is the complete scoring pattern. Notice that the same function is called three times, while each conditional updates the same score.

```python
score = 0

answer = ask_ab(
    "A mysterious door appears. You:",
    "Open it immediately",
    "Take notes before touching it"
)
if answer == "A":
    score += 1
elif answer == "B":
    score -= 1

answer = ask_ab(
    "Choose a tool.",
    "A megaphone",
    "A pocket notebook"
)
if answer == "A":
    score += 1
elif answer == "B":
    score -= 1

answer = ask_ab(
    "A plan changes at the last minute. You:",
    "Improvise and invite everyone along",
    "Find a quiet place to redesign it"
)
if answer == "A":
    score += 1
elif answer == "B":
    score -= 1

print("Final score:", score)
```

## Announce a result

Now use a branch for the final result. The labels are fictional, and the descriptions say what the quiz author means by them.

```python
if score >= 2:
    print("RESULT: Signal Fire")
    print("You like momentum, people, and making an idea visible.")
elif score <= -2:
    print("RESULT: Deep Archive")
    print("You like careful observation, focus, and building an idea over time.")
else:
    print("RESULT: Bridge")
    print("You move between spontaneous and reflective ways of working.")
```

Test the boundaries: a score of 2, 1, -1, and -2 should reach the categories you intended. The thresholds are choices made by the quiz author, not facts hidden in the data.
