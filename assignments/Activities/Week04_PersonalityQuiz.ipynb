## What's this all about?

So, personality quizzes: "Which Twin Peaks character are you?", "What kind of bread matches your soul?", "Which food cart represents your inner life?",  They're a staple of the internet, a little ridiculous, and shockingly simple to implement in code!

A personality quiz-qua-code: ask a bunch of questions, keep a tally based on the answers, and at the end announce a result based on the tally.

The reason we're doing this *now* is that a quiz runs on the big idea of this unit: **selection** — programs that *decide* things with `if`/`elif`/`else`. Every answer the player gives forces a decision: was that an A or a B? Which score gets bumped? And at the end, which result do we announce? All of that is branching, and branching is this week's skill.

A quiz also asks question after question after question, and we do *not* want to copy-paste the asking-a-question code once per question — so this activity sneaks in a **preview of next unit's big idea: loops**. ## The slow, painful way (so you appreciate the loop)

Imagine we *didn't* know about loops. To ask three questions we'd write basically the same thing three times:

```python
print("Pretend we have to do this for EVERY question...")

answer1 = input("1. Coffee or tea? ")
answer2 = input("2. Mountains or ocean? ")
answer3 = input("3. Cats or dogs? ")

print("Ugh. Imagine doing that twenty times.")
```

It works, but it's miserable. If you had twenty questions you'd copy-paste that line twenty times, and if you wanted to change how questions get asked you'd have to fix it in twenty places. There's a better way.

## The good way: a list plus a loop

The trick has two parts.

**Part one:** put all the questions in a *list* — one tidy container holding all of them in order.

**Part two:** use a `for` loop to walk through the list and do the *same thing* to each question: ask it. The loop handles "do this once for each item, until there are no items left." You write the asking code **once**.

```python
questions = [
    "Coffee or tea?",
    "Mountains or ocean?",
    "Cats or dogs?",
    "Night owl or early bird?",
]

for q in questions:
    answer = input(q + " ")
    print("You picked:", answer)

print("Done! And notice we only wrote the 'ask' part ONCE.")
```

Sit with what just happened, because it's the whole of *next* unit in four lines. The variable `q` took the value of the *first* question, ran the indented code, then took the value of the *second* question, ran the same code again, and so on — automatically — until the list ran out. **That** is iteration: the same instructions, repeated once per item, stopping on their own when the work is done. (You'll dig into loops properly next week; for now, borrowing this one is enough.)

## Keeping score

A quiz needs to *remember* something as it goes. We'll keep a running tally. Let's say every "A" answer is one kind of vibe and every "B" answer is another. We start a counter at zero before the loop, and bump it up inside the loop depending on the answer.

This is a classic pattern called an **accumulator** — a variable that lives *outside* the loop and collects results from *inside* it.

And here's where **this unit's skill** does the real work: *inside* the loop, an `if`/`else` looks at each answer and **decides** which counter to bump. Then, after the loop, another `if`/`else` decides which result to announce. The loop is just the conveyor belt; the *decisions* are the quiz.

```python
quiz = [
    "Your ideal Friday night: (A) out with a crowd  (B) home with a book — type A or B: ",
    "Your desk is usually: (A) organized chaos  (B) neat as a pin — type A or B: ",
    "A surprise day off! You: (A) make spontaneous plans  (B) finally do that project — type A or B: ",
]

a_count = 0   # the accumulators live OUT here, before the loop
b_count = 0

for question in quiz:
    response = input(question)
    if response.strip().upper() == "A":
        a_count = a_count + 1
    else:
        b_count = b_count + 1

print()
print("A answers:", a_count)
print("B answers:", b_count)

if a_count > b_count:
    print("You run on people and spontaneity.")
else:
    print("You're methodical, focused, and dependable")
```
