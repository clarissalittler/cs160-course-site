# Week 6 Activity — Build a Room Someone Else Could Enter

## What's this all about?

A whole text adventure can begin with one memorable room. In this activity, you will make a single location as a Python dictionary, display it with a triple-quoted f-string, and invite a classmate to imagine where one exit might lead.

This is a small prototype for the Week 6 assignment. You do **not** need to build movement, inventory, a command loop, or a complete game here.

## Start with one room dictionary

```python
room = {
    "title": "THE MIDNIGHT LAUNDROMAT",
    "description": """Every washing machine has stopped at exactly 11:57.
One dryer is still turning, although it is not plugged in.""",
    "exits": {"back": "alley", "down": "basement"},
    "item": "one warm red sock",
}
```

The dictionary gives names to four kinds of information. The description uses three quotation marks so it can continue across a line break.

## Turn the data into a room card

```python
exit_words = ", ".join(room["exits"].keys())

room_card = f"""
+----------------------------------------+
LOCATION: {room["title"]}
+----------------------------------------+
{room["description"]}

EXITS: {exit_words}
ITEM:  {room["item"]}
+----------------------------------------+
"""

print(room_card)
```

The labels matter more than the border. Someone should be able to understand the location, exits, and object even if the lines wrap differently on their screen.
