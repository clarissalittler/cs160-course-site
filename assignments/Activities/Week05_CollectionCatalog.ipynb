## What's this all about?

Everybody collects or keeps track of something: books, recipes, games, bus stops, favorite walks, houseplants, songs, tools, tiny rocks, or memorable snacks. This week you will turn a small collection into structured data and then ask it a question worth sharing.

One item will be a **dictionary** with named fields. The whole collection will be a **list** of those dictionaries. That list-of-records shape appears throughout real software, but here the subject and categories are yours.

Keep the data low-risk. Do not post private addresses, account information, legal names without permission, medical information, or anything another person did not agree to share.

## One item as a record

```python
item = {
    "title": "Parable of the Sower",
    "creator": "Octavia Butler",
    "category": "novel",
    "rating": 5,
}

print("Title:", item["title"])
print("Creator:", item["creator"])
print("Rating:", item["rating"])
```

The keys label each field. Before collecting data, write a tiny schema:

| Key | Type | Meaning |
|---|---|---|
| `title` | string | public title of the work |
| `creator` | string | displayed creator name |
| `category` | string | my broad, informal category |
| `rating` | integer | my rating from 1 through 5 |

The rating is an opinion, not an objective property. Naming that limitation is part of documenting data.

## Many items as a list of records

```python
collection = [
    {"title": "Parable of the Sower", "creator": "Octavia Butler", "category": "novel", "rating": 5},
    {"title": "Kindred", "creator": "Octavia Butler", "category": "novel", "rating": 5},
    {"title": "Mona Lisa Overdrive", "creator": "William Gibson", "category": "novel", "rating": 4},
    {"title": "A Wizard of Earthsea", "creator": "Ursula K. Le Guin", "category": "novel", "rating": 5},
]

print(f"Records: {len(collection)}")
for item in collection:
    print(f"{item['title']} — {item['creator']} — rating {item['rating']}")
```

Important structure is labeled in words rather than only spacing or color.

## Filter and summarize

```python
requested_creator = "Octavia Butler"
matches = []
rating_total = 0

for item in collection:
    rating_total += item["rating"]
    if item["creator"] == requested_creator:
        matches.append(item)

print(f"Average rating: {rating_total / len(collection):.1f}")
print(f"Works by {requested_creator}: {len(matches)}")
for item in matches:
    print(item["title"])
```

A filter creates a selected view; it does not delete nonmatching records. The average compresses several opinions into one number, so the report includes the record count and explains what the rating means.
